```python
import numpy as np
p = np.array([102.5, 104.0, 101.2, 105.8, 107.1])
r = np.diff(p) / p[:-1]
print( r.round(4))
print(np.diff(p))
```

```python
print(f' le rendement moyen est de {r.mean().round(3)}')
print(f" l'écart-type est de {r.std().round(3)}")
print(f' le rendement maximum est de {r.max().round(3)}')
print(f' le jour du rendement max est {r.argmax()+1}')
```

Les 3 paramètres en haut donne respectivement :
1. Le rendement moyen sur les 5 jours. Si on repartissait équitablement les rendements sur les 5 jours, chaque jour rapporterais comme rendement 1.1% . Les jours gagnants (rendement journalier positif) récompense les jours perdants (rendement journalier négatif)
2. L'écart typique d'un rendement à la moyenne, ça représente la volatilité de l'actif . Plus il est le rendement journalier est dispersé par rapport au rendement donc plus l'actif est risqué
3. Le max donne le rendement pic , le plus haut rendement journalier touché
