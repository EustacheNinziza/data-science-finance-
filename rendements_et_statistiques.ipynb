```python
import numpy as np
p = np.array([102.5, 104.0, 101.2, 105.8, 107.1])
r = np.diff(p) / p[:-1]
print( r.round(4))
print(np.diff(p))
```

```python
print(f' le rendement moyen est de {r.mean().round(3)}')
print(f" l'écart-type est de {r.std().round(3)}")
print(f' le rendement maximum est de {r.max().round(3)}')
print(f' le jour du rendement max est {r.argmax()+1}')
```

###r.mean est le rendement moyen sur les 5 jours, r.std va donner l'écart typique de chaque rendement au rendement moyen et r.max donné le rendement le plus élevé enregistré sur les 5 jours.
