```python
!pip install yfinance -q
import yfinance as yf #bibliotheque pour accéder aux données de Yahoo finance

df = yf.download("AAPL", start="2025-11-01", end="2026-11-01")
r = df["Close"].pct_change().dropna()
print(r.describe())
```
